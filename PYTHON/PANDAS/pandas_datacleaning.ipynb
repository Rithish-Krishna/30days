```python
import pandas as pd
pd.options.display.max_rows = 10

df = pd.read_csv("data2.csv")
print(df.to_string())
```

```python
#cleaning empty cell
#1. remove the row containing empty cells
new_df = df.dropna()

print(new_df.to_string())
#this produces a new dataframe 
#To change the original dataframe 
df.dropna(inplace = True)
print(new_df.to_string())
```

```python
#2 method replace the empty cell with a value 
df.fillna(130, inplace = True)

#fill the specific columns
df.fillna({"calories":130}, inplace = True)
```

```python
#3 replce with the mean median mode values
x = df["Calories"].mean()

df.fillna({"Calories": x}, inplace=True)
```
