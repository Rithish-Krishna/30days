```python
#pandas series 
# it is like a column in table 
a = [1,7,2]
my_srs = pd.Series(a)
print(my_srs)
print(my_srs[0])
```

```python
#label -> the serial no like notation before the column is called label 
#the default label is index

my_srs = pd.Series(a, index = ["x", "y", "z"])
print(my_srs)
```

```python
#dictionary as series 
#here the key becomes as index
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

```python
#We can select only specific key
myvar = pd.Series(calories, index = ["day1", "day2"])
print(myvar)
```

```python
#datafram -> Multidimensional like a table 
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data)
print(df)
```

```python
#locate a row 
#loc -> returns a series structure
print(df.loc[0])
```

```python
#loc -> with list of indexes
print(df.loc[[0,1]])
```

```python
#index naming for the dataframe 

df = pd.DataFrame(data, index = ["day1","day2","day3"]) #giving ["day1","day2"] -> result in a error
print(df)
```

```python
#locating using the named inex
print(df.loc["day1"])

print(df.loc[["day1","day3"]])
```
