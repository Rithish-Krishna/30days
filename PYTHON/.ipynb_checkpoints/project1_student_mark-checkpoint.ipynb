```python
import numpy as np

arr = np.array([32,43,54,54,65,23,64,2,3,45,89,67,34,43,67,32,56,23])
arr = np.sort(arr)
sum = np.sum(arr)
size = arr.size

median = 0
mid = 0
if(size%2==0):
    mid = int(size/2)
    median = (arr[mid-1]+arr[mid])/2
else:
    mid = size//2
    median = arr[mid]
    
mean = sum/size
print("DATABASE RESULT")
print(f"Mean mark:{int(mean)}")
print(f"Median:{median}")
print(f"Max:{arr[size-1]}")
print(f"min:{arr[0]}")
#print(f"SD:{abs(mean - (mean*mean))}")
sdarr = arr-mean
sdarr = sdarr*sdarr
sdarr = np.sum(sdarr)
print(f"SD:{np.sqrt(int(sdarr/size))}")

print(f"Toppers: {arr[arr>80]}")
print(f"Class Average:{int(mean)}")
print(f"Students below average:{arr[arr<mean]}")
```

```python
#easy method 
arr = np.array([32,43,54,54,65,23,64,2,3,45,89,67,34,43,67,32,56,23])
print("Mean   :", np.mean(arr))
print("Median :", np.median(arr))
print("Max    :", np.max(arr))
print("Min    :", np.min(arr))
print("SD     :", np.std(arr)) 
```
