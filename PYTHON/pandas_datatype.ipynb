```python
import numpy as np

#int 8,16,32,64 -> size 8bit 16bit 32bit 64 bit -> 8bit = 1byte
#default int type - int64
array = np.array([1,2,3,4,5], dtype = np.int8) 
print(array.dtype)
print(f"{array.nbytes} bytes") #8bit*5 -> 5bytes
```

```python
#float 16,32,64 
array = np.array([1,2,3,4,5], dtype = np.float16) 
print(array.dtype)
print(f"{array.nbytes} bytes") #16*5 = 80bits -> 10bytes
```

```python
#string 
array = np.array(["hello","string"], dtype = np.str_) 
print(array.dtype) #u6 -> string has 6 char -> numpy takes largest character 
print(f"{array.nbytes} bytes") # 1 char = 4byte -> u6 = 4*6 = 24 -> two string = 2*24 = 48
```

```python
#boolean
array = np.array([True,False])
print(array.dtype)
print(f"{array.nbytes} bytes") #each bool = 1 byte
```

```python
#object
array = np.array(["string",1], dtype = np.object_)
print(array.dtype)
print(f"{array.nbytes} bytes") #each element takes 8bytes
```
